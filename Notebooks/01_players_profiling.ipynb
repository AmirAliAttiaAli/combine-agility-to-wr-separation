### Draft Pick Consistency Validation

**Objective:** Verify the logical consistency between draft round, pick within round, and overall draft pick.

**Validation Rules:**
- Drafted players must have complete draft selection information.
- Draft pick values must be positive integers.
- `draft_overall_pick` must be greater than or equal to `draft_pick_within_round`.
- For Round 1, the overall pick should equal the pick within the round.

**Purpose:** Detect inconsistent draft selection records before data preprocessing.

```python
drafted = df[df["draft_round"].notna()]

invalid_picks = drafted[
    (
        drafted["draft_overall_pick"]
        < drafted["draft_pick_within_round"]
    )
    |
    (
        (drafted["draft_round"] == 1)
        &
        (
            drafted["draft_overall_pick"]
            != drafted["draft_pick_within_round"]
        )
    )
]

print("Invalid Draft Picks:", len(invalid_picks))

invalid_picks[
    [
        "display_name",
        "draft_round",
        "draft_pick_within_round",
        "draft_overall_pick"
    ]
]
```

### Birth Date Validation

**Objective:** Assess the completeness, data type, and validity of the `birth_date` column.

**Validation Checks:**
- Count missing birth dates.
- Inspect the current data type.
- Convert values to datetime for validation without modifying the original dataset.
- Identify invalid or unparseable date values.

**Purpose:** Ensure that player birth dates are reliable for future age-related analysis and feature engineering.

```python
# Check missing values
print("Missing Birth Dates:", df["birth_date"].isna().sum())

# Check current data type
print("Birth Date Data Type:", df["birth_date"].dtype)

# Convert to datetime for validation
birth_dates = pd.to_datetime(
    df["birth_date"],
    errors="coerce"
)

# Check invalid dates (excluding original missing values)
invalid_dates = (
    df["birth_date"].notna() &
    birth_dates.isna()
)

print("Invalid Birth Dates:", invalid_dates.sum())

# Check date range
print("Earliest Birth Date:", birth_dates.min())
print("Latest Birth Date:", birth_dates.max())
```

### Birth Date Validation — Findings

- All 510 players have non-null birth dates.
- No invalid or unparseable dates were detected.
- The earliest birth date is December 12, 1997.
- The latest birth date is August 25, 2004.
- The original `birth_date` column is stored as a string.

**Data Quality Conclusion:** Birth dates are complete and parseable. Date conversion is recommended during preprocessing, and player age plausibility should be validated before using age-related features.

### Player Age Plausibility Validation

**Objective:** Evaluate whether players' birth dates are plausible relative to their NFL Draft year.

**Method:** Calculate an approximate draft-year age by subtracting the birth year from the draft year.

**Validation:** Flag players whose approximate age falls outside an exploratory range of 18–30 years.

**Purpose:** Identify potentially unrealistic birth dates or draft-year values requiring further investigation.

**Note:** The age range is an initial profiling threshold, not an official NFL eligibility rule. Flagged records require review rather than automatic removal.

```python
# Calculate approximate age during draft year
approx_draft_age = (
    df["draft_year"] - birth_dates.dt.year
)

# Check age distribution
print(approx_draft_age.describe())

# Identify potentially unrealistic ages
invalid_age_mask = (
    (approx_draft_age < 18) |
    (approx_draft_age > 30)
)

print("Potentially Unrealistic Ages:", invalid_age_mask.sum())

# Display flagged records
df.loc[
    invalid_age_mask,
    ["display_name", "birth_date", "draft_year"]
]
```

### Player Age Plausibility — Findings

- Approximate draft-year age was calculated for all 510 players.
- The average approximate age is 23.11 years, with a standard deviation of 1.05.
- Player ages range from 21 to 27 years.
- The median approximate age is 23 years.
- No players fall outside the exploratory plausibility range of 18–30 years.

**Data Quality Conclusion:** No implausible ages were identified using the selected validation threshold. Birth dates appear suitable for age-related analysis, subject to any further source-level validation.

### WR Draft Status Analysis

**Objective:** Analyze the distribution of drafted and undrafted Wide Receivers.

**Method:** Classify WR players based on the availability of `draft_round`.

- **Drafted:** Players with a non-null `draft_round`.
- **Undrafted:** Players with a null `draft_round`.

**Purpose:** Understand the draft composition of the WR population and identify expected missing values in draft-related attributes.

**Note:** Missing draft information for undrafted players is expected and should not be treated as a data quality error.

```python
# Classify WR players by draft status
wr_players["draft_status"] = (
    wr_players["draft_round"]
    .apply(lambda x: "Undrafted" if pd.isna(x) else "Drafted")
)

# Count drafted vs undrafted WRs
wr_draft_status = (
    wr_players["draft_status"]
    .value_counts()
)

print(wr_draft_status)
```
